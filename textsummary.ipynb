```python
from transformers import pipeline

summarizer = pipeline("summarization", model="facebook/bart-large-cnn")
```

```python

ARTICLE = "Scene 1: Child asks Guruji (Ashram Scene) Prompt:An animated Indian child wearing a yellow kurta, sitting cross-legged under a banyan tree in a serene ashram. The child looks curious and excited, facing an elderly guru in saffron robes with long white beard, sitting calmly. Gentle golden sunlight filters through the leaves, birds chirping in the background, peaceful setting."
summary = (summarizer(ARTICLE, max_length=130, min_length=30, do_sample=False))[0]['summary_text']

import pandas as pd

summaryDf = pd.DataFrame({'summary' : [summary]})

summaryDf.to_csv('summary.csv', index=False)

# print(summary)
```
